## 1 · Iniciar un pequeño programa
**Problema.** Queremos mostrar un mensaje de inicio cuando ejecutamos un script de ventas, pero no al importar sus funciones desde otro programa. Define `main()` y ejecútala solo si el archivo es el programa principal.

**Concepto.** En un `.py`, `if __name__ == '__main__':` separa el punto de entrada de las funciones reutilizables. En Colab la celda se ejecuta al pulsar ▶; para comprobar el comportamiento al importar haría falta un archivo `.py`.


```python
def main() -> None:
    print('Iniciando el ejemplo de acceso a datos')

if __name__ == '__main__':  # No llama a main() cuando un .py se importa.
    main()
```

**Comenta:** ¿qué ocurriría al importar desde otro archivo un módulo que contuviera este código?

## 2 · Revisar una ficha de cliente incompleta
**Problema.** Recibimos una ficha con identificador, nombre, importe, confirmación y correo ausente. Debemos mostrar el tipo de cada dato y comprobar si «no hay correo» significa lo mismo que cero o texto vacío.

**Concepto.** Un nombre Python hace referencia a un objeto; `type()` muestra su tipo. `str` representa texto Unicode; UTF-8 es una codificación posible al convertir texto en bytes (Tema 02). `None` representa ausencia y no equivale a `0` ni a `''`.


```python
# Un dict relaciona cada nombre de campo con su valor.
registro = {'id': 101, 'nombre': 'Ana', 'importe': 12.50, 'confirmado': True, 'email': None}

for campo, valor in registro.items():
    print(f'{campo:<11} {str(valor):<8} {type(valor).__name__}')

# None expresa un dato ausente; no es lo mismo que un cero ni que texto vacío.
print('None == 0:', registro['email'] == 0)
print('None == cadena vacía:', registro['email'] == '')
```

**Comenta:** ¿qué diferencia hay entre una compra de importe `0` y una compra cuyo importe falta?

## 3 · Conservar los datos originales y quitar ciudades repetidas
**Problema.** Una lista de importes se entrega a otro fragmento de código: queremos comprobar si sus cambios afectan al original y cómo crear una copia independiente. También queremos obtener las ciudades únicas de un listado con repeticiones.

**Concepto.** `list` es modificable y `set` elimina duplicados (sin garantizar el orden). `alias = ventas` comparte la lista; `.copy()` hace una copia superficial (los objetos anidados mutables todavía podrían compartirse). La ficha anterior ya usa un `dict`; otras colecciones se introducirán cuando hagan falta.


```python
ventas = [12.5, 20.0]  # list: secuencia ordenada y modificable.
alias = ventas  # Otro nombre para la MISMA lista; no crea una copia.
alias[0] = 15.0
print('Lista original tras cambiar el alias:', ventas)

copia = ventas.copy()  # Copia la lista exterior; los objetos anidados pueden compartirse.
copia[0] = 99.0
print('Original:', ventas, '| Copia:', copia)

ciudades = ['Barcelona', 'Girona', 'Barcelona']
ciudades_unicas = set(ciudades)  # set: no admite duplicados.
print('Ciudades únicas:', sorted(ciudades_unicas))  # sorted ordena la salida.
```

**Comenta:** ¿por qué el cambio en `alias` afecta a `ventas`, pero el de `copia` no? ¿En qué caso `.copy()` no sería suficiente?

## 4 · Mostrar un recibo legible
**Problema.** Debemos mostrar el nombre del cliente, el total y una tabla con dos productos. Los productos deben quedar alineados a la izquierda y los importes deben tener dos decimales.

**Concepto.** `sep` controla la separación entre argumentos; `end` controla el final del `print`. Las f-strings permiten insertar valores y especificar ancho, alineación y decimales.


```python
print('Ana', 'García', sep='_')  # sep sustituye el espacio entre argumentos.
print('Total:', end=' ')  # end evita el salto de línea final.
print(f'{35.5:.2f} €')  # .2f: dos cifras decimales.

lineas = [('Teclado', 12.5), ('Ratón', 23.0)]
print(f'{"Producto":<12}|{"Importe (€)":>11}')
for producto, importe in lineas:
    print(f'{producto:<12}|{importe:>11.2f}')  # Alineación y decimales.
```

**Comenta:** predice el efecto de sustituir `end=' '` por `end='\n'`.

## 5 · Clasificar importes y categorías
**Problema.** Las ventas pueden llegar sin importe, con importe negativo o con un valor válido. Necesitamos etiquetar cada situación para revisar los datos; también traducimos categorías de productos conocidas a una descripción legible.

**Concepto.** `if`/`elif` expresan decisiones; `in` comprueba pertenencia y `match`/`case` selecciona alternativas (Python 3.10+). `case _` captura categorías no contempladas.


```python
def estado_importe(importe: float | None) -> str:
    if importe is None:  # Primero descartamos la ausencia antes de comparar números.
        return 'pendiente de revisión'
    if importe < 0:
        return 'importe inválido'
    return 'válido'

def etiqueta_categoria(categoria: str) -> str:
    match categoria:
        case 'Hardware':
            return 'Equipo físico'
        case 'Software':
            return 'Licencia o aplicación'
        case _:  # Cualquier categoría no contemplada arriba.
            return 'Otra categoría'

for importe in (None, -2.0, 0.0):
    print(importe, '→', estado_importe(importe))
print('Software' in ('Hardware', 'Software'))  # in comprueba pertenencia.
print(etiqueta_categoria('Servicios'))
```

**Comenta:** ¿por qué se comprueba `importe is None` antes de comparar con cero?

## 6 · Contar y sumar registros con un dato ausente
**Problema.** Se han recibido cuatro importes; uno falta. Hay que identificar qué número de registro debe revisarse y calcular el total de los importes presentes, sin confundir ausencia con cero.

**Concepto.** `enumerate(..., start=1)` aporta número de registro y valor; `continue` omite el importe ausente y sigue con el siguiente registro. Un bucle `while` se verá en el ejemplo interactivo, donde no sabemos cuántas veces habrá que pedir el dato.


```python
importes = [12.5, None, 0.0, 9.75]
validos = []
for numero, importe in enumerate(importes, start=1):  # Posición y valor.
    if importe is None:
        print(f'Registro {numero}: falta el importe')
        continue  # Salta SOLO este registro y sigue con el siguiente.
    validos.append(importe)  # Un cero es válido según la regla de este ejemplo.

print('Registros con importe:', len(validos))
print('Importes válidos:', validos)
print('Total:', sum(validos))
```

**Comenta:** ¿qué cambiaría en el total si la política de datos exigiera rechazar los importes cero? `continue` pasa a la siguiente vuelta del bucle; no termina el bucle entero.

## 7 · Crear una función reutilizable para el total
**Problema.** Queremos calcular totales de diferentes listas sin repetir la fórmula y comprobar dos casos: una lista vacía y dos ventas con total conocido.

**Concepto.** `def` define una función; `return` entrega el resultado. Las anotaciones (`list[float]`, `-> float`) explican qué espera y qué devuelve, pero Python no las comprueba automáticamente al ejecutar. `assert` verifica el resultado esperado.


```python
def importe_total(importes: list[float]) -> float:
    """Calcula la suma de los importes recibidos."""
    return sum(importes)  # return entrega el resultado a quien llamó a la función.

print(f'Total calculado: {importe_total([12.5, 9.75]):.2f} €')
assert importe_total([]) == 0  # También se comprueba la lista vacía.
assert importe_total([12.5, 9.75]) == 22.25
```

**Comenta:** ¿por qué se prueba también una lista vacía? `assert` sirve aquí para una comprobación sencilla; para suites de proyecto se usan herramientas específicas.

## 8 · Interpretar importes recibidos como texto
**Problema.** Un sistema entrega `' 12.50 '`, `'N/A'` y `'0'` como cadenas. Hay que obtener un número cuando sea posible e identificar el dato que no puede convertirse, sin sustituirlo por cero.

**Concepto.** `strip()` elimina espacios; `float()` interpreta el número y `except ValueError` captura únicamente una conversión fallida. El valor rechazado no se convierte por ello en válido: el Tema 02 mostrará cómo registrar rechazos al leer ficheros.


```python
def convertir_importe(texto: str) -> float | None:
    try:
        return float(texto.strip())  # strip elimina espacios; float convierte texto.
    except ValueError:  # Solo capturamos el error de conversión esperado.
        return None  # El dato rechazado no se sustituye silenciosamente por cero.

for texto in (' 12.50 ', 'N/A', '0'):
    importe = convertir_importe(texto)
    print(f'{texto!r} → {importe!r}')

assert convertir_importe('N/A') is None
```

**Comenta:** ¿por qué sería peligroso sustituir `N/A` directamente por `0`? ¿Qué error se produciría si `texto` fuese `None` en vez de `str`?


## 8b · Pedir un importe al usuario
**Problema.** El programa debe pedir un importe no negativo; si el usuario escribe `N/A`, texto inválido o un número negativo, debe volver a pedirlo. En modo demostración recibe `'12.50'` para que «Ejecutar todo» no quede esperando.

**Concepto.** `input(mensaje)` pausa el programa y **siempre entrega `str`**; la función anterior convierte el texto, y `while` permite reintentar hasta obtener un dato admitido. Para usar el teclado pon `MODO_INTERACTIVO = True` y ejecuta **solo esta celda**. No introduzcas datos personales.


```python
MODO_INTERACTIVO = False  # True activa el teclado; False permite «Ejecutar todo».

if MODO_INTERACTIVO:
    while True:
        texto = input('Introduce un importe no negativo: ')  # Devuelve str.
        importe = convertir_importe(texto)  # Convertir puede fallar.
        if importe is None or importe < 0:
            print('Dato inválido: escribe un número no negativo.')
            continue  # Vuelve a preguntar; no procesa el dato inválido.
        break
else:
    texto = '12.50'  # Dato ficticio para ejecutar el cuaderno sin interacción.
    importe = convertir_importe(texto)

print(f'Entrada recibida: {texto!r} (tipo {type(texto).__name__})')
print(f'Importe validado: {importe:.2f} €')
```

**Comenta:** ¿por qué `input()` entrega `'12.50'` como texto? ¿Qué sucede al introducir `N/A` o un número negativo en modo interactivo?


## 9 · Impedir importes negativos en una venta
**Problema.** Una venta debe guardar su identificador y un importe no negativo. Si se intenta registrar un importe negativo al crearla o modificarla, debe lanzar un error; además, queremos mostrarla de forma legible.

**Concepto.** `__init__` inicializa la instancia y `self` la representa. `_importe` indica uso interno por convenio. `@property` permite leer `venta.importe`, el setter comprueba una asignación y `__str__` define qué muestra `print(venta)`.


```python
class Venta:
    def __init__(self, identificador: int, importe: float) -> None:
        self.identificador = identificador
        self.importe = importe  # Pasa por el setter: también valida el valor inicial.

    @property  # Hace posible leer venta.importe sin paréntesis.
    def importe(self) -> float:
        return self._importe

    @importe.setter  # Intercepta venta.importe = nuevo_valor.
    def importe(self, nuevo: float) -> None:
        if nuevo < 0:
            raise ValueError('El importe no puede ser negativo')
        self._importe = nuevo  # El guion bajo indica uso interno por convenio.

    def __str__(self) -> str:
        return f'Venta {self.identificador}: {self.importe:.2f} €'

venta = Venta(101, 12.50)
print(venta)  # Invoca __str__ para producir el texto.
venta.importe = 15.00  # El setter comprueba el nuevo importe.
print(venta)
try:
    venta.importe = -1
except ValueError as error:
    print('Asignación rechazada:', error)
```

**Comenta:** ¿qué ventaja tiene validar desde `__init__` y no solo cuando cambiamos el importe después? ¿Por qué no hacen falta `getImporte()` y `setImporte()` al estilo Java?

## 10 · Añadir una categoría a un registro existente
**Problema.** Todos los registros tienen un identificador. Los registros de venta, además, necesitan una categoría. Queremos reutilizar la inicialización del identificador y comprobar que la venta también es un registro. Es un ejemplo pequeño de herencia, no una recomendación de crear una jerarquía para cualquier fila.

**Concepto.** La clase hija amplía una clase base; `super()` ejecuta su inicializador y `isinstance()` comprueba la relación entre instancia y clase.


```python
class Registro:
    def __init__(self, identificador: int) -> None:
        self.identificador = identificador

class RegistroVenta(Registro):  # Hereda de Registro.
    def __init__(self, identificador: int, categoria: str) -> None:
        super().__init__(identificador)  # Inicializa la parte heredada.
        self.categoria = categoria

fila = RegistroVenta(101, 'Hardware')
print(fila.identificador, fila.categoria)
print('Es un Registro:', isinstance(fila, Registro))  # True por herencia.
assert issubclass(RegistroVenta, Registro)
```
